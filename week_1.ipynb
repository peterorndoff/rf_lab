```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv("example_data_1_PHYS_433_533.csv")
```

```python
print(df)
```

```python
frequency = df['f_res']
power = df['max_power_dBm']
```

```python
plt.title("Power vs Frequency")
plt.xlabel('Frequency')
plt.ylabel('Power')
plt.plot(frequency, power)
```

```python
power_mid = 10 ** (np.array(power) / 10)
print(power)
```

```python
plt.title("Power vs Frequency")
plt.xlabel('Frequency')
plt.ylabel('Power')
plt.plot(frequency, power_mid)
```

```python
mask = frequency > 4000

frequency_filtered = frequency[mask]
power_filtered = power_mid[mask]
plt.title("Power vs Frequency")

plt.xlabel('Frequency')
plt.ylabel('Power')
plt.plot(frequency_filtered, power_filtered)
```
